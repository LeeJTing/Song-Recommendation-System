# Data Preparation

### Data Loading

```python
import numpy as np
import pandas as pd

df1 = pd.read_csv('data/Music Info.csv')
df1.head()
```

```python
df2 = pd.read_csv('data/User Listening History.csv')
df2.head()
```

### Selected only the relevant features

```python
df1_selected = df1[['track_id','name','artist','spotify_preview_url','spotify_id','tags','genre','year','danceability','loudness','speechiness','tempo','time_signature']]
df1_selected.head()
```

```python
df2_selected = df2[['track_id','user_id','playcount']]
df2_selected.head()
```

### Merge two Data Frames

```python
merged_df = pd.concat([df1_selected,df2_selected],ignore_index = True)
merged_df.head()
```

### Check data type

```python
merged_df.dtypes
```

### Save the data frame to reviews.csv to proceed data preprocessing

```python
merged_df.to_csv("data/reviews.csv",index=False)
print("selected dataset save as reviews.csv")
```
