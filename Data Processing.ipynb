# Data Processing

### Data Loading

```python
import numpy as np
import pandas as pd

df = pd.read_csv('data/reviews.csv',low_memory=False)
df.head()
```

### Clean null and duplicates value in the dataset

```python
df.dropna(subset=["tags","genre"],inplace=True)
df.head()
```

```python
# Clean track_id: strip spaces, convert to lowercase
df['track_id'] = df['track_id'].astype(str).str.strip().str.lower()

# Remove duplicate track_id, keeping only the first one
df.drop_duplicates(subset='track_id', keep='first', inplace=True)
df.head()
```

### Save the data frame to data.csv to proceed data preprocessing

```python
df.to_csv("data/data.csv", index=False)
print("Merged dataset saved as data.csv")
```
