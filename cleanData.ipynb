# Cleaning Datasets


## Process:
1. Load the Dataset
2. Handle Missing Values
3. Remove Duplicates
4. Clean Text Data
5. Save the Cleaned Dataset

# Import Library

```python
import pandas as pd
```

# 1. Loading The Dataset

```python
# Music Information
music_df = pd.read_csv('data/MusicInfo.csv')

# User Listening History
history_df = pd.read_csv('data/UserListeningHistory.csv')

print("Number Of Row(Music Information): ", len(music_df))
print("Number Of Row(User Listening History): ", len(history_df))
```

# 2. Remove useless column

```python
music_df.dropna(inplace=True)  # Drop rows with any missing values
music_df.drop(columns=['spotify_id', 'energy', 'key', 'loudness', 'mode', 'acousticness', 'instrumentalness', 'liveness', 'valence', 'time_signature'], inplace=True)
```

# 3. Handle Missing Values

```python
# Detect missing values
print("Music Information: ")
print(music_df.isnull().sum())

print("\nUser Listening History: ")
print(history_df.isnull().sum())

history_df.dropna(inplace=True)  # Drop rows with any missing values
# User History: Remove rows for track IDs that no longer exist in the music info file
valid_track_ids = set(music_df['track_id'])
history_df = history_df[history_df['track_id'].isin(valid_track_ids)]

print("\nNumber Of Row(Music Information): ", len(music_df))
print("Number Of Row(User Listening History): ", len(history_df))
```

# 4. Remove Duplicates

```python
music_df.drop_duplicates(inplace=True)
history_df.drop_duplicates(inplace=True)

print("\nNumber Of Row(Music Information): ", len(music_df))
print("Number Of Row(User Listening History): ", len(history_df))
```

# 5. Clean Text Data

```python
# Music Name
music_df['name'] = music_df['name'].str.strip()        # Remove leading/trailing spaces

# Artist
music_df['artist'] = music_df['artist'].str.strip() 
music_df['artist'] = music_df['artist'].str.upper()

# Music Tags
music_df['tags'] = music_df['tags'].str.strip() 
music_df['tags'] = music_df['tags'].str.upper()

# Genre
music_df['genre'] = music_df['genre'].str.strip() 
music_df['genre'] = music_df['genre'].str.upper()

# Track ID
music_df['track_id'] = music_df['track_id'].str.strip()
music_df['track_id'] = music_df['track_id'].str.upper()
history_df['track_id'] = history_df['track_id'].str.strip()
history_df['track_id'] = history_df['track_id'].str.upper()

# User ID
history_df['user_id'] = history_df['user_id'].str.strip()
history_df['user_id'] = history_df['user_id'].str.upper()
print("Used Music Fields:",music_df.head())
print("Used User Listening History Fields:", history_df.head())
```

# 6. Simplify User ID

```python
# Create mapping from unique user_id to USER1, USER2,...
user_mapping = {uid: f'USER{i+1}' for i, uid in enumerate(history_df['user_id'].unique())}

# Replace old IDs with new ones
history_df['user_id'] = history_df['user_id'].map(user_mapping)
print(history_df.head(5))
```

# 7. Save the Cleaned Dataset

```python
music_df.to_csv('data/newMusicInfo.csv', index=False)
history_df.to_csv('data/newUserListeningHistory.csv', index=False)
```
